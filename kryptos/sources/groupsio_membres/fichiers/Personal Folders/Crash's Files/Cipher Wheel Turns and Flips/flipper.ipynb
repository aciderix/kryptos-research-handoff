```python
def alignClearZenithLookup(ct, pt, pt_char):
    ct_char=None
    for i in range(0,len(pt)):
        pos=-1
        if pt_char==pt[i]:
            pos=i
            ct_char=ct[i]
            break
    if not pos==-1:
        rotated_ct=ct[pos:] + ct[:pos]
        rotated_pt=pt[pos:] + pt[:pos]
    else:
        rotated_ct=ct
        rotated_pt=pt
    return rotated_ct,rotated_pt,ct_char
```

```python
def alignCodeZenithLookup(ct, pt, ct_char):
    pt_char=None
    for i in range(0,len(ct)):
        pos=-1
        if ct_char==ct[i]:
            pos=i
            pt_char=pt[i]
            break
    if not pos==-1:
        rotated_ct=ct[pos:] + ct[:pos]
        rotated_pt=pt[pos:] + pt[:pos]
    else:
        rotated_ct=ct
        rotated_pt=pt
    return rotated_ct,rotated_pt,pt_char
```

```python
def codeLookup(ct,pt,ct_char):
    pt_char=None
    for i in range(0,len(ct)):
        if ct_char==ct[i]:
            pt_char=pt[i]
            break
    
    return pt_char
```

```python
def alignCTPTPairAtZenith(ct,pt,ct_char,pt_char):
    for i in range(0,len(ct)):
        pos=-1
        if ct_char==ct[i]:
            pos=i
            break
    if not pos==-1:
        rotated_ct=ct[pos:] + ct[:pos]
    else:
        rotated_ct=ct
        
    for i in range(0,len(pt)):
        pos=-1
        if pt_char==pt[i]:
            pos=i
            break
    if not pos==-1:
        rotated_pt=pt[pos:] + pt[:pos]
    else:
        rotated_pt=pt
    return rotated_ct,rotated_pt
```

```python
def alignClearZenithAxis(pt, pt_char):

    for i in range(0,len(pt)):
        pos=-1
        if pt_char==pt[i]:
            pos=i
            break
    if not pos==-1:
        rotated_pt=pt[pos:] + pt[:pos]
    else:
        rotated_pt=pt
    return rotated_pt
```

```python
def rotatePT(pt,n):
    
    # n is -25 to +25
    
    rotated_pt=pt[n:] + pt[:n]
    
    return rotated_pt
```

```python
def zenithFlipPT(pt):
    
    # flips PT wheel over aligned to zenith - rotate first with alignClearZenithLookup
    
    rotated_pt=[pt[13]]+pt[12:0:-1]+[pt[0]]+pt[26:13:-1]
    
    return rotated_pt
```
